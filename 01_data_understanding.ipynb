```python
clean_df = df.drop_duplicates().copy()  #new dataset which delets all duplicated elements

print("Original rows:", len(df))
print("Rows after removing duplicates:", len(clean_df))
print("Duplicates remaining:", clean_df.duplicated().sum())
```

```python
clean_df["Sales"] = clean_df["Quantity"] * clean_df["UnitPrice"]

clean_df[["Quantity", "UnitPrice", "Sales"]].head(10)
```

```python
print("Cleaned dataset shape:", clean_df.shape)
print("Total sales value:", clean_df["Sales"].sum())
```

```python
clean_df["IsCancelled"] = (
    clean_df["InvoiceNo"].astype(str).str.startswith("C")
)

clean_df["IsReturn"] = clean_df["Quantity"] < 0

print("Cancelled invoice rows:", clean_df["IsCancelled"].sum())
print("Negative quantity rows:", clean_df["IsReturn"].sum())
```

```python
print("Total rows:", len(clean_df))    #new rows added iscancelled and isreturned

print(
    "Rows with both cancellation and negative quantity:",
    (clean_df["IsCancelled"] & clean_df["IsReturn"]).sum()
)

print(
    "Negative quantity rows without C invoice number:",
    ((~clean_df["IsCancelled"]) & clean_df["IsReturn"]).sum()
)
```

```python
sales_df = clean_df[~clean_df["IsCancelled"]].copy()

print("Rows for initial sales analysis:", len(sales_df))
print("Cancelled rows remaining:", sales_df["IsCancelled"].sum())  #main thing
print("Total sales value:", sales_df["Sales"].sum())
```

```python
sales_df["InvoiceDate"] = pd.to_datetime(sales_df["InvoiceDate"])

sales_df["YearMonth"] = (sales_df["InvoiceDate"].dt.to_period("M"))
sales_df[["InvoiceDate", "YearMonth", "Sales"]].head()

#pd.to_datetime() ensures the dates are in the correct format.
#.dt lets us extract information from dates.
#.to_period("M") groups dates by month and year, such as 2010-12.
```

```python
monthly_sales = (sales_df.groupby("YearMonth")["Sales"].sum())

monthly_sales
#adds sales value of every month
```

```python
print("Highest sales month:")
print(monthly_sales.idxmax(), monthly_sales.max())

print("\nLowest sales month:")
print(monthly_sales.idxmin(), monthly_sales.min())

#.idxmax() returns the month with the highest sales value.
#.max() returns that month's sales value.
#.idxmin() and .min() do the same for the lowest value
```

```python
#monthly sales chart 
monthly_sales.index = monthly_sales.index.astype(str)

plt.figure(figsize=(12, 6))
monthly_sales.plot(kind="line", marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales Value (£)")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
#checks customer information
customer_df = sales_df.dropna(subset=["CustomerID"]).copy()

print("Total sales rows:", len(sales_df))
print("Rows with customer IDs:", len(customer_df))
print("Unique customers:", customer_df["CustomerID"].nunique())
print("Missing customer IDs:", customer_df["CustomerID"].isnull().sum())
```

```python
#find your most unique customer
customer_orders = (customer_df.groupby("CustomerID")["InvoiceNo"].nunique().sort_values(ascending=False))

customer_orders.head(10)
```

```python
#calculate customer spending
customer_spending = (customer_df.groupby("CustomerID")["Sales"].sum().sort_values(ascending=False))

customer_spending.head(10)
```

```python
#comspre cutomer frequency spending
customer_summary = customer_df.groupby("CustomerID").agg(TotalOrders=("InvoiceNo", "nunique"),TotalSpending=("Sales", "sum"),TotalItems=("Quantity", "sum"))

customer_summary.head()
```

```python
customer_summary.sort_values("TotalSpending", ascending=False).head(10)
```

```python
customer_summary.sort_values( "TotalOrders", ascending=False).head(10)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

plt.scatter(
    customer_summary["TotalOrders"],
    customer_summary["TotalSpending"],
    alpha=0.5
)

plt.title("Customer Spending vs Order Frequency")
plt.xlabel("Number of Orders")
plt.ylabel("Total Spending (£)")
plt.tight_layout()
plt.show()
```
